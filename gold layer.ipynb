```python
silver_df=spark.table("silver_sales_ecom")
```

```python
silver_df.printSchema()
```

```python
display(silver_df.limit(10))
```

## Customer Analytics

```python
from pyspark.sql.functions import sum,avg,countDistinct
gold_customer_analytics=silver_df.groupBy('customer_unique_id')\
                   .agg((sum('payment_value').alias('total_spend')),
                        avg('payment_value').alias('avg_order_value'),
                        countDistinct('order_id').alias('total_orders'))
```

```python
display(gold_customer_analytics.limit(10))
```

```python
gold_customer_analytics.write.format('delta').mode("overwrite").saveAsTable("gold_customer_analytics")
```

## Product Analytics

```python
from pyspark.sql.functions import sum,countDistinct

gold_product_analytics = silver_df.groupBy(
    "product_category_name"
).agg(
    sum("payment_value").alias("total_revenue"),
    countDistinct("order_id").alias("total_orders")
)
```

```python
gold_product_analytics.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold_product_analytics")
```

## Payment Analytics

```python
gold_payment_analytics = silver_df.groupBy(
    "payment_type"
).agg(
    countDistinct("order_id").alias("total_orders"),
    sum("payment_value").alias("total_revenue")
)
```

```python
gold_payment_analytics.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold_payment_analytics")
```

## Monthly Revenue Trend

```python
gold_monthly_revenue = silver_df.groupBy(
    "order_year",
    "order_month"
).agg(
    sum("payment_value").alias("monthly_revenue")
)
```

```python
gold_monthly_revenue.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold_monthly_revenue")
```

## Executive KPI Dashboard

```python
from pyspark.sql.functions import sum,countDistinct,avg

gold_kpi_dashboard = silver_df.agg(
    sum("payment_value").alias("total_revenue"),
    countDistinct("order_id").alias("total_orders"),
    countDistinct("customer_unique_id").alias("total_customers"),
    avg("payment_value").alias("average_order_value")
)
```

```python
gold_kpi_dashboard.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold_kpi_dashboard")
```
