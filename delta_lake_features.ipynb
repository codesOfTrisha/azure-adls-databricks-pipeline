```python
%sql
DESCRIBE DETAIL silver_sales_ecom
```

### Update

```python
%sql
update silver_sales_ecom
set customer_city='Bangalore'
where customer_city='Sao Paulo'
```

```python
%sql
SELECT *
FROM silver_sales_ecom
WHERE customer_city='Bangalore'
LIMIT 10;
```

### Delete

```python
spark.sql("DELETE FROM default.silver_sales_ecom WHERE customer_city='Bangalore'")
```

```python
%sql
SELECT *
FROM silver_sales_ecom
WHERE customer_city='Bangalore';
```
