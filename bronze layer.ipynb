# Customer

```python
df_customer=spark.read.format('csv')\
                      .option('header','true')\
                      .option('inferSchema','true')\
                      .load('/Volumes/databricksrajanya/default/raw_files/olist_customers_dataset.csv')
```

```python
df_customer.printSchema()
```

```python
df_customer.count()
```

```python
display(df_customer.limit(10))
```

```python
df_customer.write.format('delta')\
                 .mode('overwrite')\
                 .saveAsTable('bronze_customer')
```
