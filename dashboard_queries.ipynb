### Total Revenue KPI

```python
%sql
SELECT
ROUND(SUM(total_revenue),2) AS Total_Revenue
FROM gold_product_analytics;
```

## Total Customer

```python
%sql
select count (*) as Total_Customer from gold_customer_analytics
```

### Average Order Value

```python
%sql
select round(avg(avg_order_value),2) as Avg_Order_value from gold_customer_analytics
```

### Top Product Categories

```python
%sql
SELECT
product_category_name,
total_revenue
FROM gold_product_analytics
ORDER BY total_revenue DESC
LIMIT 10;
```

### Payment Type Analysis

```python
%sql
SELECT
payment_type,
SUM(total_revenue) as total_revenue
FROM gold_payment_analytics
GROUP BY payment_type
ORDER BY total_revenue DESC;
```
